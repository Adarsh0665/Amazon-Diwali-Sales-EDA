# Unit Price vs Total Sales

```python
hist = ['Unit_Price_INR','Total_Sales_INR']
for col in hist:
    plt.subplot(1,2,hist.index(col)+1)
    sns.histplot(sales[col],kde=True )
plt.show()
```

# Product Data Distribution

```python
columns = ['Quantity', 'Unit_Price_INR', 'Total_Sales_INR', 'Review_Rating']
for col in columns:
    plt.figure(figsize=(4,2))
    sns.histplot(sales[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
category_col = ['Product_Category', 'Payment_Method', 'Delivery_Status', 'State']
for col in category_col:
    plt.figure(figsize=(7,3))
    sns.countplot(data=sales, x=col, order=sales[col].value_counts().index)
    plt.title(f'{col} Distribution')
    plt.xticks(rotation=90)
    plt.show()
```

# Top Products BY Sales

```python
Top_Products = sales.groupby('Product_Name')['Total_Sales_INR'].sum().sort_values(ascending=False).head(20)
sns.barplot(x=Top_Products.index, y=Top_Products.values)
plt.ylabel("Total Sales (INR)")
plt.xlabel("Product Name")
plt.yscale("log")
plt.xticks(rotation=90)
plt.title("Top 20 Products by Total Sales")
plt.show()
```

# Total Sales by Category

```python
plt.figure(figsize=(8,4))
sns.barplot(data=sales, x='Product_Category', y='Total_Sales_INR', estimator=sum)
plt.title('Total Sales by Product Category')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.barplot(data=sales, x='Product_Category', y='Review_Rating')
plt.title('Customer Ratings by Category')
plt.show()
```

# Monthly Sales Trend

```python
sales['Date'] = pd.to_datetime(sales['Date'])
sales['Month'] = sales['Date'].dt.month_name()

plt.figure(figsize=(10,4))
sns.barplot(data=sales, x='Month', y='Total_Sales_INR', estimator=sum)
plt.title('Monthly Sales Trend')
plt.xticks(rotation=90)
plt.show()
```

# Sales by Payment Method

```python
payment = sales.groupby('Payment_Method')['Total_Sales_INR'].sum().sort_values(ascending=False)
sns.barplot(x=payment.index, y=payment.values,)
plt.ylabel("Total Sales (INR)")
plt.xlabel("Payment Method")
plt.xticks(rotation=90)
plt.title("Total Sales by Payment Method")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.barplot(data=sales, x='Payment_Method', y='Total_Sales_INR', estimator='mean')
plt.title('Rating vs Delivery Status')
plt.xticks(rotation=90)
plt.show()
```

# Return Data

```python
returns_category = sales.groupby('Product_Category')['Delivery_Status'].apply(lambda x: (x=='Returned').mean()*100)
returns_category.sort_values(ascending=False).plot(kind='bar', figsize=(8,4))
plt.title('Return Rate by Product Category (%)')
plt.ylabel('Return %')
plt.show()
```

```python
returns_category = sales.groupby('State')['Delivery_Status'].apply(lambda x: (x=='Returned').mean()*100)
returns_category.sort_values(ascending=False).plot(kind='bar', figsize=(8,4))
plt.title('Return Rate by State (%)')
plt.ylabel('Return %')
plt.show()
```

```python
returns_category = sales.groupby('Review_Rating')['Delivery_Status'].apply(lambda x: (x=='Returned').mean()*100)
returns_category.sort_values(ascending=True).plot(kind='bar', figsize=(8,4))
plt.title('Return Rate by ratings (%)')
plt.ylabel('Return %')
plt.xticks(rotation=0)
plt.show()
```

# Rating by ProductName 

```python
plt.figure(figsize=(10,4))
sns.barplot(data=sales, x='Product_Name', y='Review_Rating', estimator='mean', order=sales.groupby('Product_Name')['Review_Rating'].mean().sort_values(ascending=False).index)
plt.title('Average Rating by ProductName')
plt.xticks(rotation=90)
plt.show()
```

# Average Rating by State

```python
plt.figure(figsize=(10,4))
sns.barplot(data=sales, x='State', y='Review_Rating', estimator='mean', order=sales.groupby('State')['Review_Rating'].mean().sort_values(ascending=False).index)
plt.title('Average Rating by State')
plt.xticks(rotation=90)
plt.show()
```
